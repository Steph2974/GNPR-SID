```python
import pandas as pd
import ast

def colision(codebook):
    """
    计算 codebook 的冲突情况。
    
    Args:
        codebook: codebook.csv 的文件路径 (str) 或已读取的 DataFrame
    """
    if isinstance(codebook, str):
        try:
            df = pd.read_csv(codebook)
        except FileNotFoundError:
            print(f"File not found: {codebook}")
            return
    else:
        df = codebook
    
    if 'Codebook' not in df.columns:
        print("Error: DataFrame must contain a 'Codebook' column.")
        return

    # 提取所有码字的前三位 (原始 Semantic ID)
    # Codebook 列存储的是字符串形式的列表，例如 "[24, 57, 54, 0]"
    # 列表前三位是 RQ-VAE 生成的原始码字，第四位（如果有）是冲突解决序号
    raw_codes = []
    for item in df['Codebook']:
        try:
            # 安全解析字符串列表，兼容已经是 list 的情况
            code_list = ast.literal_eval(item) if isinstance(item, str) else item
            # 截取前 3 位作为原始 ID
            original_code = tuple(code_list[:3])
            raw_codes.append(original_code)
        except Exception as e:
            continue

    total_count = len(raw_codes)
    unique_set = set(raw_codes)
    unique_count = len(unique_set)
    collision_count = total_count - unique_count
    collision_rate = collision_count / total_count if total_count > 0 else 0.0

    print(f"Total Samples: {total_count}")
    print(f"Unique Count: {unique_count}")
    print(f"Collision Count: {collision_count}")
    print(f"Collision Rate: {collision_rate:.6f}")

    return unique_count, collision_count, collision_rate
```

```python
unique_count, collision_count, collision_rate = colision("datasets/TKY/codebooks_v0_0.5.csv")
```

```python
codebooks = {
    "datasets/NYC-v00/codebooks_v00_0.5.csv",
    "datasets/NYC-v0/codebooks_v0_0.5.csv",
    "datasets/NYC-v0/codebooks_v0.3_0.5.csv",
    "datasets/NYC-v1/codebooks_v1_0.5.csv"
}
for codebook in codebooks:
    print(f"***** {codebook.split('/')[-1]} *****")
    unique_count, collision_count, collision_rate = colision(codebook)
    # print(f"Unique Count: {unique_count}")
    # print(f"Collision Count: {collision_count}")
    # print(f"Collision Rate: {collision_rate:.6f}")
```
